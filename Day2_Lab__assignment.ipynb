```python
def add_student_record(registry):
    """Task 1: Prompt user for student ID, name, branch, and CGPA.

    Validation Rules:
      1. Student ID must not already exist in the registry.
      2. Name and Branch must not be empty after stripping whitespace.
      3. CGPA must be a valid float between 0.0 and 10.0.
      4. Auto-generate email: <first_name_lowercase>.<id_lowercase>@university.edu
    """
    print("\n--- Add New Student ---")

    s_id = input("Enter Student ID: ").strip()
    if not s_id:
        print("[ERROR] Student ID cannot be empty.")
        return
    if s_id in registry:
        print(f"[ERROR] Student ID '{s_id}' already exists.")
        return

    name = input("Enter Name: ").strip()
    branch = input("Enter Branch: ").strip()

    if not name or not branch:
        print("[ERROR] Name and Branch cannot be empty.")
        return

    try:
        cgpa = float(input("Enter CGPA (0.0-10.0): "))
        if not (0.0 <= cgpa <= 10.0):
            print("[ERROR] CGPA must be between 0.0 and 10.0.")
            return
    except ValueError:
        print("[ERROR] CGPA must be a valid number.")
        return

    # Auto-generate email: first_name.id@university.edu
    first_name = name.split()[0].lower()
    email = f"{first_name}.{s_id.lower()}@university.edu"

    # Addition to registry dictionary
    registry[s_id] = {
        "name": name,
        "branch": branch,
        "cgpa": cgpa,
        "email": email
    }

    print(f"[SUCCESS] Record for {name} added successfully.")
```

```python
def search_student_record(registry):
    """Task 2: Search by student ID (exact) or student Name (case-insensitive substring)."""
    print("\n--- Search Student Records ---")
    query = input("Enter Student ID or Name to search: ").strip().lower()

    if not query:
        print("[ERROR] Search query cannot be empty.")
        return

    results = []
    for s_id, details in registry.items():
        # Check for exact ID match or name substring match
        if query == s_id.lower() or query in details['name'].lower():
            results.append((s_id, details))

    if not results:
        print(f"[INFO] No records found matching '{query}'.")
    else:
        print(f"\nFound {len(results)} match(es):")
        print("-" * 60)
        for s_id, details in results:
            print(f"ID: {s_id} | Name: {details['name']} | Branch: {details['branch']} | CGPA: {details['cgpa']}")
        print("-" * 60)
```

```python
def delete_student_record(registry):
    """Task 3: Prompt for Student ID and delete record with confirmation."""
    print("\n--- Delete Student Record ---")
    s_id = input("Enter Student ID to delete: ").strip()

    if s_id in registry:
        student_name = registry[s_id]['name']
        confirm = input(f"Are you sure you want to delete the record for '{student_name}' (ID: {s_id})? (yes/no): ").strip().lower()

        if confirm == 'yes':
            del registry[s_id]
            print(f"[SUCCESS] Student record '{s_id}' has been deleted.")
        else:
            print("[INFO] Deletion cancelled.")
    else:
        print(f"[ERROR] Student ID '{s_id}' not found in registry.")
```

```python
def update_student_record(registry):
    """Task 4: Implement update_student_record() to modify existing student details."""
    print("\n--- Update Student Record ---")
    s_id = input("Enter Student ID to update: ").strip()

    if s_id not in registry:
        print(f"[ERROR] Student ID '{s_id}' not found.")
        return

    student = registry[s_id]
    print(f"Updating records for {student['name']}. Leave blank to keep current value.")

    new_name = input(f"New Name [{student['name']}]: ").strip()
    new_branch = input(f"New Branch [{student['branch']}]: ").strip()
    new_cgpa_str = input(f"New CGPA [{student['cgpa']}]: ").strip()

    if new_name:
        student['name'] = new_name
        # Update email if name changes
        first_name = new_name.split()[0].lower()
        student['email'] = f"{first_name}.{s_id.lower()}@university.edu"

    if new_branch:
        student['branch'] = new_branch

    if new_cgpa_str:
        try:
            new_cgpa = float(new_cgpa_str)
            if 0.0 <= new_cgpa <= 10.0:
                student['cgpa'] = new_cgpa
            else:
                print("[WARN] Invalid CGPA range. Keeping old value.")
        except ValueError:
            print("[WARN] Invalid CGPA format. Keeping old value.")

    print(f"[SUCCESS] Record for ID '{s_id}' updated.")
```

```python
import json

def save_records_to_json(file_path, registry):
    """Task 5: Implement save_records_to_json() with safe file flushing.
    Serializes the registry dictionary to a JSON file.
    """
    print(f"\n--- Saving Records to {file_path} ---")
    try:
        with open(file_path, "w", encoding="utf-8") as file:
            json.dump(registry, file, indent=2)
            file.flush()
        print(f"[SUCCESS] {len(registry)} record(s) saved successfully to '{file_path}'.")
    except Exception as e:
        print(f"[ERROR] Failed to save records: {e}")
```

```python
import csv

def export_to_csv(file_path, registry):
    """Task 6 (Stretch Goal): Implement export_to_csv().
    Exports all student records to a CSV file using DictWriter.
    """
    if not registry:
        print("[WARN] No records found to export.")
        return

    print(f"\n--- Exporting Records to {file_path} ---")
    try:
        # Define the CSV headers based on keys + student ID
        fieldnames = ['id', 'name', 'branch', 'cgpa', 'email']

        with open(file_path, "w", newline='', encoding="utf-8") as csvfile:
            writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
            writer.writeheader()

            for s_id, details in registry.items():
                # Create a row dictionary combining the ID and the other details
                row = {'id': s_id}
                row.update(details)
                writer.writerow(row)

        print(f"[SUCCESS] Data successfully exported to '{file_path}'.")
    except Exception as e:
        print(f"[ERROR] Export failed: {e}")
```
