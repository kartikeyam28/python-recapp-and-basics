```python
# 1. Takes student name, age, and marks in 3 subjects as input.
name = input("Enter student's name: ")
age = int(input("Enter student's age: "))

mark1 = float(input("Subject 1 marks: "))
mark2 = float(input("Subject 2 marks: "))
mark3 = float(input("Subject 3 marks: "))
marks = [mark1, mark2, mark3]

# 2. Displays all entered details using print().
print("\n Student Details ")
print(f"Name: {name}")
print(f"Age: {age}")
print(f"Marks in 3 subjects: {marks}")

# 3. Calculates total marks, average, and percentage.

total_marks = sum(marks)
average = total_marks / len(marks)
percentage = (total_marks / 300) * 100

print("\n Academic Performance ")
print(f"Total Marks: {total_marks}")
print(f"Average: {average: }")
print(f"Percentage: {percentage: }%")

# 4. Checks whether the student scored 40+ in every subject AND has an average of 60+

passed = all(m >= 40 for m in marks)
eligible_average = average >= 60

print("\n Merit Program Eligibility ")
if passed and eligible_average:
    print("Status: Eligible for Merit Program")
else:
    print("Status: Not Eligible for Merit Program")

#Displays the data type of name, age, marks, and average.
print("\n Data Types ")
print(f"DT 'name': {type(name)}")
print(f"DT 'age': {type(age)}")
print(f"DT 'marks': {type(marks)}")
print(f"DT 'average': {type(average)}")
```

```python
n = int(input("Enter a number: "))

count = 0

for i in range(1, n + 1):
    if n % i == 0:
        count += 1

if count == 2:
    print("Prime number")
else:
    print("Not a prime number")
```

```python
n = int(input("Enter a number: "))

sum_even = 0
sum_odd = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        sum_even += i
    else:
        sum_odd += i

print(f"Sum of even numbers up to {n}: {sum_even}")
print(f"Sum of odd numbers up to {n}: {sum_odd}")
```

```python
n = int(input("Enter a number: "))

fact = 1

for i in range(1, n + 1):
    fact = fact * i

print("Factorial =", fact)
```
